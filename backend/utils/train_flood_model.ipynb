```python
import pandas as pd
rain = pd.read_csv("../data/rainfall_processed.csv")
elev = pd.read_csv("../data/elevation_processed.csv")
rain.head()
```

```python
elev.head()
```

```python
rain = rain.rename(columns={"state_name":"state"})
```

```python
flood_data = elev.copy()
# temporary rainfall mapping
flood_data["actual_rainfall"] = rain["actual"].mean()
flood_data["rain_deviation"] = rain["deviation"].mean()
flood_data["rain_risk"] = rain["rain_risk"].mean()

flood_data.head()
```

```python
flood_data["flood_risk_score"] = (
    flood_data["terrain_risk"] + flood_data["rain_risk"]
)
```

```python
def flood_label(score):
    if score >= 3:
        return 3
    elif score >= 2:
        return 2
    elif score >= 1:
        return 1
    else:
        return 0

flood_data["flood_risk"] = flood_data["flood_risk_score"].apply(flood_label)
```

```python
features = flood_data[[
    "latitude",
    "longitude",
    "elevation",
    "terrain_risk",
    "rain_risk"
]]

target = flood_data["flood_risk"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.2,
    random_state=42
)

model = RandomForestClassifier(n_estimators=200)

model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score
pred = model.predict(X_test)
accuracy_score(y_test, pred)
```

```python
import joblib
joblib.dump(model,"../models/flood_ml_model.pkl")
joblib.dump(features.columns.tolist(),"../models/flood_model_features.pkl")
```

```python
import pandas as pd
data = pd.read_csv("../data/elevation_processed.csv")
data.head()
```

```python
import numpy as np

micro_points = []

for _, row in data.iterrows():
    for i in range(5):   # 5 micro hotspots per district
        lat = row["latitude"] + np.random.uniform(-0.05,0.05)
        lon = row["longitude"] + np.random.uniform(-0.05,0.05)

        micro_points.append({
            "district": row["district"],
            "latitude": lat,
            "longitude": lon,
            "terrain_risk": row["terrain_risk"]
        })

micro_hotspots = pd.DataFrame(micro_points)

micro_hotspots.head()
```

```python
micro_hotspots.shape
```

```python
from sklearn.cluster import KMeans
features = micro_hotspots[["latitude","longitude","terrain_risk"]]
kmeans = KMeans(n_clusters=60, random_state=42)
micro_hotspots["cluster"] = kmeans.fit_predict(features)
micro_hotspots.head()
```

```python
zones = micro_hotspots.groupby("cluster").agg({
    "terrain_risk":"mean",
    "district":"count"
}).reset_index()
zones.rename(columns={"district":"hotspot_count"}, inplace=True)
zones.head()
```

```python
zones["readiness_score"] = 100 - (zones["terrain_risk"] * 30)
```

```python
micro_hotspots.to_csv("../data/india_flood_micro_hotspots.csv", index=False)
zones.to_csv("../data/flood_zone_summary.csv", index=False)
```
